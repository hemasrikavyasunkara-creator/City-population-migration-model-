```python
import numpy as np
import matplotlib.pyplot as plt

print("City Population Migration Model is ready!")
```

```python
def get_float(prompt, default):
    value = input(f"{prompt} [{default}]: ").strip()
    if value == "":
        return float(default)
    return float(value)


def get_int(prompt, default):
    value = input(f"{prompt} [{default}]: ").strip()
    if value == "":
        return int(default)
    return int(value)


print("Enter the initial population of the three cities.")
print("Press Enter to use the default value.\n")

A0 = get_float("City A population", 1000)
B0 = get_float("City B population", 1000)
C0 = get_float("City C population", 1000)

years = get_int("Number of years", 10)

print("\nEnter migration percentages.")
print("Example: 10 means 10% of the city's population migrates.\n")

A_to_B = get_float("A to B (%)", 10)
A_to_C = get_float("A to C (%)", 10)

B_to_A = get_float("B to A (%)", 10)
B_to_C = get_float("B to C (%)", 10)

C_to_A = get_float("C to A (%)", 10)
C_to_B = get_float("C to B (%)", 10)

# Basic validation
if A0 < 0 or B0 < 0 or C0 < 0:
    raise ValueError("Population cannot be negative.")

if years <= 0:
    raise ValueError("Number of years must be greater than 0.")

percentages = [A_to_B, A_to_C, B_to_A, B_to_C, C_to_A, C_to_B]

if any(p < 0 or p > 100 for p in percentages):
    raise ValueError("Migration percentages must be between 0 and 100.")

if A_to_B + A_to_C > 100:
    raise ValueError("Total migration from City A cannot exceed 100%.")

if B_to_A + B_to_C > 100:
    raise ValueError("Total migration from City B cannot exceed 100%.")

if C_to_A + C_to_B > 100:
    raise ValueError("Total migration from City C cannot exceed 100%.")

print("\nInputs accepted successfully!")
```

```python
# Convert percentages into decimal proportions
ab = A_to_B / 100
ac = A_to_C / 100

ba = B_to_A / 100
bc = B_to_C / 100

ca = C_to_A / 100
cb = C_to_B / 100

# Migration matrix
# Rows = destination cities
# Columns = source cities

M = np.array([
    [1 - ab - ac, ba,             ca],
    [ab,          1 - ba - bc,    cb],
    [ac,          bc,             1 - ca - cb]
], dtype=float)

# Initial population vector
P0 = np.array([
    [A0],
    [B0],
    [C0]
], dtype=float)

print("Migration Matrix M:")
print(M)

print("\nInitial Population Vector P(0):")
print(P0)

print("\nMatrix shape:", M.shape)
print("Population vector shape:", P0.shape)
```

```python
print("Population using Matrix Powers")

for year in range(1, years + 1):
    P = M @ P
    print("Year", year, ":", P.flatten())
```

```python
eigenvalues, V = np.linalg.eig(M)

D = np.diag(eigenvalues)

print("Eigenvalues:")
print(eigenvalues)

print("\nMatrix V:")
print(V)

print("\nDiagonal Matrix D:")
print(D)

print("\nM = VDV^-1")
```

```python
P = np.array([[A], [B], [C]])

A_values = [A]
B_values = [B]
C_values = [C]

for year in range(years):
    P = M @ P
    A_values.append(P[0, 0])
    B_values.append(P[1, 0])
    C_values.append(P[2, 0])

plt.plot(A_values, label="City A")
plt.plot(B_values, label="City B")
plt.plot(C_values, label="City C")

plt.xlabel("Year")
plt.ylabel("Population")
plt.title("City Population Migration")

plt.legend()
plt.grid()
plt.show()
```

```python
print("Final Population")

print("City A:", round(A_values[-1], 2))
print("City B:", round(B_values[-1], 2))
print("City C:", round(C_values[-1], 2))
```

```python
test_P = np.array([[1000], [1000], [1000]])

test_result = M @ test_P

print("Test Case:")
print("Initial population = 1000, 1000, 1000")

print("\nAfter 1 year:")
print(test_result.flatten())

print("\nExpected:")
print("[1000, 1000, 1000]")
```
